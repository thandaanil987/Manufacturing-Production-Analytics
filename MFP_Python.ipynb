```python
# date cleaning

df.columns=(df.columns.str.strip().str.lower().str.replace(" ","_"))

df['production_date']=pd.to_datetime(df['production_date'],errors='coerce')

df=df.drop_duplicates()
```

```python

df["defect_rate"] = (
    df["defective_units"]
    .div(df["units_produced"].where(df["units_produced"] != 0))
    * 100
)

print(df[["units_produced", "defective_units", "defect_rate"]].head())
```

```python
# saving updated dataset 

df.to_csv("Production_Dataset.csv")
```
